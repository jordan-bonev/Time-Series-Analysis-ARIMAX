## 1.2. Test for stationarity - ADF

```python
def adf_test(df, date_col=None, signif=0.05):
    """
    Perform Augmented Dickey-Fuller (ADF) test on multiple time series.

    Parameters:
        df : pandas DataFrame
            Columns are time series; optionally one date column
        date_col : str or None
            Name of the date column to ignore
        signif : float
            Significance level for p-value threshold (default 0.05)

    Returns:
        pandas DataFrame with ADF test results and classification
    """
    results = {}

    for col in df.columns:
        if col == date_col:
            continue
        series = df[col].dropna()
        stat, pval, lags, nobs, crit_vals, icbest = adfuller(series, autolag='AIC')

        # Classification based on p-value
        if pval < signif:
            classification = "Stationary"
        else:
            classification = "Non-stationary"

        results[col] = {
            "ADF Statistic": stat,
            "p-value": pval,
            "Lags Used": lags,
            "Num Observations": nobs,
            "1% Critical": crit_vals['1%'],
            "5% Critical": crit_vals['5%'],
            "10% Critical": crit_vals['10%'],
            "Classification": classification
        }

    return pd.DataFrame(results).T
```

## 1.3. KPSS test for stationarity

```python

def kpss_test(df, date_col=None, regression='c', nlags='auto'):
    """
    Perform KPSS test on multiple time series and determine stationarity.

    Parameters:
        df : pandas DataFrame
            Columns are time series; optionally one date column
        date_col : str or None
            Name of the date column to ignore
        regression : {'c','ct'}
            'c' for constant (level stationarity), 'ct' for constant + trend (trend stationarity)
        nlags : int or 'auto'
            Number of lags to use in the test

    Returns:
        pandas DataFrame with KPSS results and classification
    """
    results = {}

    for col in df.columns:
        if col == date_col:
            continue
        series = df[col].dropna()
        stat, pval, lags, crit_vals = kpss(series, regression=regression, nlags=nlags)

        # Classification based on p-value
        if pval > 0.05:
            classification = "Stationary"
        elif pval <= 0.05 and pval > 0.01:
            classification = "Probably non-stationary (weak evidence)"
        else:
            classification = "Non-stationary"

        results[col] = {
            "KPSS Statistic": stat,
            "p-value": pval,
            "Lags Used": lags,
            "10% Critical": crit_vals['10%'],
            "5% Critical": crit_vals['5%'],
            "2.5% Critical": crit_vals['2.5%'],
            "1% Critical": crit_vals['1%'],
            "Classification": classification
        }

    return pd.DataFrame(results).T
```

## 1.4 Test for Trend

```python
def check_trend_mk(df, date_col=None):
    """
    Check for trend in each time series using the Mann-Kendall test.
    
    Parameters:
        df : pandas DataFrame
            Columns are time series; optionally one date column
        date_col : str or None
            Name of the date column to ignore
            
    Returns:
        pandas DataFrame with:
            - Trend type ('increasing', 'decreasing', 'no trend')
            - Test statistic
            - p-value
            - Significance (True if p < 0.05)
    """
    results = {}

    for col in df.columns:
        if col == date_col:
            continue
        series = df[col].dropna()
        mk_result = mk.original_test(series)
        
        results[col] = {
            'Trend': mk_result.trend,
            'Test Statistic': mk_result.slope,
            'p-value': mk_result.p,
            'Significant': mk_result.p < 0.05
        }
    
    return pd.DataFrame(results).T
```

## 1.5 Test for Seasonality

```python
def check_seasonality(df, date_col=None, period=None):
    """
    Check for seasonality in each time series.

    Parameters:
        df : pandas DataFrame
            Columns are time series; optionally one date column
        date_col : str or None
            Name of the date column to ignore
        period : int or None
            Expected seasonal period (e.g., 12 for monthly data with yearly seasonality)
    
    Returns:
        pandas DataFrame with:
            - Seasonal presence (True/False)
            - Strength of seasonality (variance ratio: seasonal/total)
    """
    results = {}

    for col in df.columns:
        if col == date_col:
            continue
        series = df[col].dropna()

        # Decompose series if period is given
        if period is None:
            # Try to estimate period from ACF peaks (not implemented automatically here)
            seasonal_strength = None
            has_seasonality = None
        else:
            decomposition = seasonal_decompose(series, model='additive', period=period, extrapolate_trend='freq')
            seasonal = decomposition.seasonal
            residual = decomposition.resid

            # Strength of seasonality: variance ratio
            var_seasonal = seasonal.var()
            var_residual = residual.var()
            seasonal_strength = var_seasonal / (var_seasonal + var_residual)

            # Arbitrary threshold: >0.1 indicates seasonality
            has_seasonality = seasonal_strength > 0.1

        results[col] = {
            'Seasonal Strength': seasonal_strength,
            'Seasonality Detected': has_seasonality
        }

    return pd.DataFrame(results).T
```

```python
seasonality_results = check_seasonality(train, date_col='Date', period=12)
seasonality_results
```

## 1.6. Analyze Seasonality

```python
def analyze_seasonality(df, date_col, period):
    """
    Analyze seasonality for multiple time series using STL decomposition + ACF + seasonal strength.

    Parameters:
        df : pandas DataFrame
            Must include a datetime column and one or more time series columns
        date_col : str
            Name of the date column
        period : int
            Expected seasonality period (e.g., 12 for monthly data with yearly cycle)
    """
    df = df.copy()
    df[date_col] = pd.to_datetime(df[date_col])
    df = df.set_index(date_col)

    summary = {}

    for col in df.columns:
        print(f"Analyzing: {col}")

        series = df[col].dropna()

        # STL decomposition
        stl = STL(series, period=period, robust=True)
        res = stl.fit()

        # Plot STL components
        fig = res.plot()
        fig.set_size_inches(12, 8)
        fig.suptitle(f"{col} - STL Decomposition (Period={period})", fontsize=14, y=1.02)
        plt.tight_layout()
        plt.show()

        # ACF plot (lags cannot exceed series length-1)
        max_lag = min(len(series) - 1, period * 2)
        plt.figure(figsize=(10,4))
        plot_acf(series, lags=max_lag)
        plt.title(f"{col} - Autocorrelation (ACF)")
        plt.tight_layout()
        plt.show()

        # Compute seasonal strength using variance ratio (robust)
        seasonal = res.seasonal
        trend = res.trend
        resid = res.resid

        total_var = np.var(trend + seasonal + resid)
        strength = np.var(seasonal) / total_var if total_var > 0 else np.nan

        summary[col] = strength
        print(f"Seasonal Strength ({col}): {strength:.3f}")

    # Return a summary DataFrame
    return pd.DataFrame(summary, index=["Seasonal Strength"]).T
```

## 1.7. Make stationary

```python
def make_stationary(df, log_cols=None, diff_cols=None, seasonal_diff_cols=None, seasonal_period=12):
    """
    Transform time series to stationary using log, first difference, and seasonal difference.

    Parameters:
        df : pandas DataFrame
            Input data with time series columns
        log_cols : list of str or None
            Columns to apply log transform
        diff_cols : list of str or None
            Columns to apply first difference (trend removal)
        seasonal_diff_cols : list of str or None
            Columns to apply seasonal differencing (seasonality removal)
        seasonal_period : int
            Period for seasonal differencing (e.g., 12 for monthly data)
    
    Returns:
        pandas DataFrame with transformed series
    """
    df_transformed = df.copy()
    
    # 1. Log transform
    if log_cols is not None:
        for col in log_cols:
            # Avoid log(0) issues
            df_transformed[col] = np.log(df_transformed[col].replace(0, 1e-8))
    
    # 2. First differencing
    if diff_cols is not None:
        for col in diff_cols:
            df_transformed[col] = df_transformed[col].diff()
    
    # 3. Seasonal differencing
    if seasonal_diff_cols is not None:
        for col in seasonal_diff_cols:
            df_transformed[col] = df_transformed[col] - df_transformed[col].shift(seasonal_period)
    
    # Drop initial NaNs created by differencing
    df_transformed = df_transformed.dropna().reset_index(drop=True)
    
    return df_transformed
```

# 2. CCF and Granger Causality

## 2.1. CCF

```python
def check_ccf_statsmodels(main_series, exog_df, max_lag=None, plot=True):
    results = {}
    main_series = main_series.dropna()

    for col in exog_df.columns:
        exog_series = exog_df[col].dropna()
        combined = pd.concat([main_series, exog_series], axis=1).dropna()
        x = combined.iloc[:, 0]
        y = combined.iloc[:, 1]

        # Determine safe max_lag
        available_lags = len(x) - 1
        if max_lag is None or max_lag > available_lags:
            max_lag_use = available_lags
        else:
            max_lag_use = max_lag

        ccf_values = ccf(y, x)[:max_lag_use+1]
        lags = range(0, max_lag_use + 1)

        results[col] = ccf_values

        if plot:
            plt.figure(figsize=(8, 4))
            plt.stem(lags, ccf_values)
            plt.xlabel('Lag')
            plt.ylabel('Cross-correlation')
            plt.title(f'CCF (statsmodels): {col} -> main series')
            plt.show()

    return pd.DataFrame(results, index=lags)
```

# 2.2. Granger Causality

```python
def granger_causality(df, main_col, exog_cols, max_lag=6, alpha=0.05):
    """
    Run Granger causality tests and highlight significant p-values.

    Parameters:
        df : pandas DataFrame
            DataFrame containing stationary time series
        main_col : str
            Column name of the main series (dependent)
        exog_cols : list of str
            Column names of exogenous series
        max_lag : int
            Maximum number of lags to test
        alpha : float
            Significance threshold for highlighting (default 0.05)

    Returns:
        pandas Styler object: table of p-values with significant cells highlighted in green
    """
    results = []

    # Combine main + exogenous
    all_cols = [main_col] + exog_cols
    data = df[all_cols].dropna()

    for lag in range(1, max_lag + 1):
        row = {'Lag': lag}

        # Main causing itself
        try:
            test_main = grangercausalitytests(data[[main_col, main_col]], maxlag=lag, verbose=False)
            pval_main = test_main[lag][0]['ssr_chi2test'][1]  # p-value
        except:
            pval_main = None
        row[f'{main_col} -> {main_col}'] = pval_main

        # Exogenous causing main
        for exog in exog_cols:
            try:
                test_exog = grangercausalitytests(data[[main_col, exog]], maxlag=lag, verbose=False)
                pval_exog = test_exog[lag][0]['ssr_chi2test'][1]  # p-value
            except:
                pval_exog = None
            row[f'{exog} -> {main_col}'] = pval_exog

        results.append(row)

    df_results = pd.DataFrame(results)

    # Styling function
    def highlight_significant(val):
        if pd.notnull(val) and val < alpha:
            return 'background-color: lightgreen'
        return ''

    return df_results.style.applymap(highlight_significant)
```

```python
main = 'AAPL'
exog = ['TSLA', 'BTC_USD', 'KO', 'MCD']

granger_table = granger_causality(train_stationary, main, exog, max_lag=2)
granger_table
```

```python
def select_exog_lags(main_series, exog_df, max_granger_lag=12, max_ccf_lag=12, plot=False, significance=0.05):
    """
    Automatically select exogenous variables and their best lag using Granger causality + CCF.

    Parameters:
        main_series : pd.Series
            The main dependent series
        exog_df : pd.DataFrame
            Candidate exogenous variables
        max_granger_lag : int
            Max lag to test in Granger causality
        max_ccf_lag : int
            Max lag to consider in CCF
        plot : bool
            Whether to plot CCF for significant variables
        significance : float
            p-value threshold for Granger causality

    Returns:
        pd.DataFrame with columns: ['variable', 'best_lag', 'ccf_value']
    """
    selected = []

    main_series = main_series.dropna()

    for col in exog_df.columns:
        exog_series = exog_df[col].dropna()
        combined = pd.concat([main_series, exog_series], axis=1).dropna()
        x = combined.iloc[:, 0]  # main
        y = combined.iloc[:, 1]  # exogenous

        # --- 1. Granger causality ---
        significant = False
        try:
            test_result = grangercausalitytests(combined[[x.name, y.name]], maxlag=max_granger_lag, verbose=False)
            for lag, res in test_result.items():
                p_value = res[0]['ssr_chi2test'][1]  # p-value from chi2 test
                if p_value < significance:
                    significant = True
                    break
        except Exception as e:
            print(f"Granger test failed for {col}: {e}")
            continue

        if not significant:
            continue  # skip variable if not Granger-causal

        # --- 2. Cross-correlation ---
        available_lags = len(x) - 1
        max_lag_use = min(max_ccf_lag, available_lags)
        ccf_values = ccf(y, x)[:max_lag_use+1]
        lags = np.arange(0, max_lag_use+1)
        best_idx = np.argmax(np.abs(ccf_values))
        best_lag = lags[best_idx]
        best_ccf = ccf_values[best_idx]

        selected.append({'variable': col, 'best_lag': best_lag, 'ccf_value': best_ccf})

        if plot:
            plt.figure(figsize=(8,4))
            plt.stem(lags, ccf_values)
            plt.xlabel('Lag')
            plt.ylabel('CCF')
            plt.title(f'{col} -> {x.name} | Best lag: {best_lag}')
            plt.show()

    return pd.DataFrame(selected).sort_values('best_lag')
```

```python
selected_exog = select_exog_lags(
    main_series=train_stationary['AAPL'],
    exog_df=train_stationary[['TSLA','BTC_USD','KO','MCD']],
    max_granger_lag=2,
    max_ccf_lag=11,
    plot=True
)
print(selected_exog)
```

# 3. ARIMAX

```python
def prepare_exog(df, selected_exog):
    """
    Shift exogenous variables by their chosen lag and return aligned DataFrame.
    """
    exog_aligned = pd.DataFrame(index=df.index)
    
    for _, row in selected_exog.iterrows():
        var = row['variable']
        lag = row['best_lag']
        exog_aligned[var] = df[var].shift(lag)
    
    # Drop rows with NaN (caused by shifting)
    exog_aligned = exog_aligned.dropna()
    return exog_aligned
```

```python
exog_data = prepare_exog(train_stationary, selected_exog)

# Align main series
main_aligned = train_stationary['AAPL'].loc[exog_data.index]
```

```python
model = SARIMAX(
    main_aligned,
    exog=exog_data,
    order=(1, 1, 1),      # (p,d,q)
    seasonal_order=(0,0,0,0), 
    enforce_stationarity=False,
    enforce_invertibility=False
)

model_fit = model.fit(disp=False)
print(model_fit.summary())
```

```python
# Forecast next 12 periods
forecast_exog = exog_data[-12:]  
forecast = model_fit.get_forecast(steps=6, exog=forecast_exog)
pred_mean = forecast.predicted_mean
pred_ci = forecast.conf_int()

print(pred_mean)
print(pred_ci)
```

## 3.1. Invert stationarity

```python
def invert_stationary_forecast_multiple(forecast_diff, original_series, log_cols=[], diff_cols=[], seasonal_diff_cols=[], seasonal_period=12):
    """
    Invert forecast for main series and optionally exogenous series.
    
    forecast_diff : pd.Series (stationary forecast)
    original_series : pd.Series (original, pre-stationary)
    log_cols : list of series names that were log-transformed
    diff_cols : list of series names that were first-differenced
    seasonal_diff_cols : list of series names that were seasonally differenced
    """
    forecast = forecast_diff.copy()
    
    # 1. Seasonal differencing
    if seasonal_diff_cols is not None and forecast.name in seasonal_diff_cols:
        forecast += original_series.shift(seasonal_period).iloc[-len(forecast):].values
    
    # 2. First differencing
    if diff_cols is not None and forecast.name in diff_cols:
        forecast = forecast.cumsum() + original_series.iloc[-1]
    
    # 3. Log transform
    if log_cols is not None and forecast.name in log_cols:
        forecast = np.exp(forecast)
    
    return forecast
```
